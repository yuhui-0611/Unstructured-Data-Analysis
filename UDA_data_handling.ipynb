```python
from google.colab import files
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter

# 1. 검수용 데이터 생성
drama_review = drama[
    ['video_id', 'drama_guess', 'title', 'url']
].copy()

# 2. 그룹별 영상 개수 계산
drama_review['group_count'] = (
    drama_review.groupby('drama_guess', dropna=False)['video_id']
    .transform('size')
)

# 3. 검수용 컬럼 추가
drama_review['drama_name'] = ''
drama_review['review_status'] = 'pending'

# 4. 그룹별 정렬 (영상 수 많은 그룹부터)
drama_review = drama_review.sort_values(
    ['group_count', 'drama_guess'],
    ascending=[False, True],
    kind='stable'
).reset_index(drop=True)

# 5. 엑셀 컬럼 순서 정리
drama_review = drama_review[
    ['drama_guess', 'group_count', 'video_id',
     'title', 'url', 'drama_name', 'review_status']
]

# 6. 엑셀 저장
filename = 'drama_review.xlsx'
drama_review.to_excel(filename, index=False)

# 7. 엑셀 서식 설정
wb = load_workbook(filename)
ws = wb.active
ws.title = '드라마명 검수'

# 제목 행 스타일
for cell in ws[1]:
    cell.font = Font(bold=True, color='FFFFFF')
    cell.fill = PatternFill('solid', fgColor='263B59')
    cell.alignment = Alignment(horizontal='center')

# 유튜브 URL 클릭 가능하게 설정
for row in range(2, ws.max_row + 1):
    cell = ws.cell(row=row, column=5)
    if cell.value:
        cell.hyperlink = cell.value
        cell.font = Font(color='0563C1', underline='single')

# 컬럼 너비
widths = [28, 15, 20, 55, 48, 28, 18]
for i, width in enumerate(widths, 1):
    ws.column_dimensions[get_column_letter(i)].width = width

# 첫 행 고정 + 필터
ws.freeze_panes = 'A2'
ws.auto_filter.ref = ws.dimensions

wb.save(filename)

# 8. 결과 확인 및 다운로드
print('전체 영상:', len(drama_review))
print('드라마명 그룹:', drama_review['drama_guess'].nunique(dropna=False))

display(drama_review.head(10))

files.download(filename)
```

```python
import pandas as pd

# 원본 메타데이터
original = drama.copy()

# 검수 완료한 엑셀
reviewed = pd.read_excel('/content/drama_review_final.xlsx')

# video_id 중복 및 누락 확인
assert original['video_id'].is_unique
assert reviewed['video_id'].is_unique

# 검수한 컬럼만 추출
reviewed = reviewed[
    ['video_id', 'drama_name', 'review_status']
]

# video_id 기준으로 조인
final = original.merge(
    reviewed,
    on='video_id',
    how='left',
    validate='one_to_one'
)

# 결과 검증
print("원본 데이터:", len(original))
print("최종 데이터:", len(final))
print("드라마명 미입력:", final['drama_name'].isna().sum())

display(final.head())
```

```python
final.to_csv(
    'shorts_metadata_final.csv',
    index=False,
    encoding='utf-8-sig'
)

from google.colab import files
files.download('shorts_metadata_final.csv')
```
